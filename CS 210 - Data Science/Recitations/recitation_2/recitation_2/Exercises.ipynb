## Q1: Z-Score Calculation



In this exercise, you are going to practise a statistical notion called `z-score` by utilizing some of the basic functionalities of the numpy library.

To describe the notion briefly, `z-score` can be considered as the measure of distance of data points from their mean scores. `z-score` of a particular data instance indicates how many mean distances that point is away the mean of the population.

If you want to study the `z-score` further, you can check following URLs.

[StatisticsHowTo.com - z-score Definition](https://www.statisticshowto.com/probability-and-statistics/z-score/)

[Khan Academy - An Exercise including z-score](https://www.youtube.com/watch?v=Wp2nVIzBsE8)

The formula below shows how the z-score for a particular data point can be calculated. Notations used in the formula are also explained in the table below.

$$ z_i = \frac{(x_i - \mu)}{\sigma} $$

| Tables        | Description             | 
| ------------- |:-------------:| 
| $z_i$     | z-score of data point i     |
| $x_i$      | value of the data point i  |  
| $\mu$ | mean value of the collection    |
| $\sigma$ | standard deviation of the collection     |
