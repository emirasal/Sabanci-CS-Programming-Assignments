# Matplotlib

Matplotlib is a Python plotting library in which you can generate plots, histograms, power spectra, bar charts, errorcharts, scatterplots, etc., with just a few lines of code.

![](https://matplotlib.org/_images/sphx_glr_scatter_demo2_0011.png)

````python
import matplotlib.pyplot as plt  # we'll be using pyplot with plt alias
import numpy as np

# graphs will be included in the notebook
%matplotlib inline
````

````python
n = 5
x_values = np.arange(n)
y_values = np.square(x_values)
#y=x^2
plt.plot(x_values, y_values,color="red",linestyle="",marker="*")
plt.show()
````

````python
# let's plot cosine and sine functions
x = np.linspace(-np.pi, np.pi, 256, endpoint=True)
c = np.cos(x) # y=cos(x)
s = np.sin(x) # y=sin(x)

plt.plot(x, c)  # first parameter x-axis, second one y-axis
plt.plot(x, s)
plt.show()
````

## Chart Elements

<img height="500" width="500" src="https://matplotlib.org/_images/anatomy.png"></img>

*Image obtained from https://matplotlib.org*

````python
# playing with the defaults
x = np.linspace(-np.pi, np.pi, 256, endpoint=True)
c = np.cos(x)
s = np.sin(x)

# blue color with a continuous line of width 1 (pixels)
plt.plot(x, c, color="blue", linewidth=1.0, linestyle=":", label="cosine")

# green color with a continuous line of width 1 (pixels)
plt.plot(x, s, color="green", linewidth=1.0, linestyle="-.", label="sine")

# set x limits
plt.xlim(-4.0,4.0)
# set y limits
plt.ylim(-1.0,1.0)

# set x ticks
plt.xticks(np.linspace(-4,4,9,endpoint=True))
#plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi])
#plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi], [r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'])

# set y ticks
plt.yticks(np.linspace(-1,1,5,endpoint=True))

# enable the legend
plt.legend()

# add title
plt.title("sine - cosine")
plt.xlabel("x values")
plt.ylabel("y values")

plt.show()
````

## Chart Types

### Scatter Plot

````python
data = {
        "a": np.arange(50),
        "c": np.random.randint(0, 50, 50),
        "d": np.abs(np.random.randn(50)) * 100
       }

# adding a new key
data["b"] = data["a"] + 10 * np.random.randn(50)
plt.scatter(data["a"], data["b"])
plt.show()
plt.scatter(data["a"], data["b"], c=data["c"], s=data["d"])  # setting x and y axes from the data
                                                             # c = color coding, s = size of each dot
plt.xlabel("a")
plt.ylabel("b")
plt.show() 
````

### Histograms

````python
x = np.random.normal(10, 5, 1000)  # normal dist. with mean set to 10 and standard deviation to 5

plt.hist(x, bins=10)
plt.show()
````

### Bar Charts

````python
x = ['Nuclear', 'Hydro', 'Gas', 'Oil', 'Coal', 'Biofuel']
energy = np.random.randint(10, 100, len(x))

plt.bar(x, energy, color='green')
plt.xlabel("Energy Source")
plt.ylabel("Energy Output (GJ)")
plt.title("Energy output from various fuel sources")

plt.show()
````

## Playing with the layout

``` py
>>> fig = plt.figure()  # create a figure object
>>> ax = fig.add_subplot(1, 1, 1)  # create an axes object in the figure
                                   # (row, column, position)
>>> ax.plot(x)  # now, we can plot on axes ax
```

``` py
>>> fig, axs = plt.subplots(1, 3)  # returns a figure in which 3 axes are placed horizontally
```

````python
data = {'apples': 10, 'oranges': 15, 'lemons': 5, 'limes': 20}
names = list(data.keys())
values = list(data.values())

fig, axs = plt.subplots(1,3, figsize=(10, 3), sharey=True)

axs[0].bar(names, values)
axs[1].scatter(names, values)
axs[2].plot(names, values)
fig.suptitle('Categorical Plotting')  # figure title
````

## Widgets

````python
from ipywidgets import interact
````

````python
x = np.linspace(-np.pi, np.pi, 256, endpoint=True)

def plotsin(f):  # parameter will be updated with the slider
    plt.plot(x, np.sin(x*f))
    plt.show()
    
interact(plotsin, f=(1, 10, 1))  # start, end, stride
````

## Exercise

- Create a horizontal bar chart with random data.
- Include the bar values as text in the figure.

````python
n = 10
values = np.random.uniform(10, 100, n)
cat_options = [chr(i) for i in range(ord('A'), ord('Z') + 1)]
categories = np.random.choice(cat_options, n, replace=False)

categories = categories[values.argsort()]
values = np.sort(values)

plt.barh(categories, values)

for value, cat in zip(values, categories):
  plt.text(value, cat, f"{value:.2f}", verticalalignment="center", fontsize=8)

plt.xlim([0, values.max()+10])
plt.xlabel("Values")
plt.ylabel("Categories")
plt.title("Bar Chart with Text")

plt.show()
````
