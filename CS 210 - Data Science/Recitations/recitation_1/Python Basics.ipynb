#### Pass by Reference or Pass by Value

In Python every variable name is a reference. When we pass a variable to a function, a new reference to the object is created.

**Small Note**

In Python, we have a function named `id`, which takes an obect as the single parameter and returns a unique and constant integer value during its life time.

``` py
>>> a = 5
>>> id(a)
94714354481696
```

````python
def print_val(x):
    """
    prints the id and value of x
    """
    print("id of x before assingment:", id(x))
    x = 4

# let's create an atomic variable
a = 5
# and print its id
print("id of a:", id(a))

# we'll see that the id of a and x are the same
# but the value of a did not change
print_val(a)
print("a:", a)
````

When we pass a reference and change the received reference to something else,   
the connection between passed and received parameter is broken.

That's what happens in `print_val` function.
If we print the id of `x` after the assignment,  
we'll see a different id value.

````python
def print_val(x):
    """
    prints the id and value of x
    """
    print("id of x before assingment:", id(x))
    x = 4
    print("id of x after assingment:", id(x))

a = 5
print("id of a:", id(a))

print_val(a)
print("a:", a)
````

When we pass non-atomic values, such as list, dictionaries etc.,  
as parameters to functions, we need to pay attention to undesired outcomes.

````python
def change_val(lst):
    """
    sets the first element in lst to zero
    """
    lst[0] = 0
    
# list creation
lst = [10, 11, 12, 13, 14, 15]
print("initial values in the list:", lst)

change_val(lst)
print("after the function call:", lst) 
````
