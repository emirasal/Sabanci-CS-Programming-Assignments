# Pandas & Regex Exercise

In this notebook, we have an exercise from the [Stack Overflow question dataset](https://www.kaggle.com/imoore/60k-stack-overflow-questions-with-quality-rate). 

The dataset contains related information and the Quality Rating of questions posted in Stack Overflow.

For each question in the dataset, the Quality Rating can be found from the corresponding `Y` attribute value. Currently, there are three possible values `Y` attribute can take. Meanings of these values are listed below:

- HQ: High-quality posts
- LQ_EDIT: Low-quality posts with a negative score, but still open
- LQ_CLOSE: Low-quality posts that were closed by the community

Our first hypothesis with this dataset was; *if the presence or the absence of selected tags in the question titles affected the Quality Rating of the question.*

To answer this question, we will make use of `regex` *(Regular Expressions)* and a function of Pandas called `crosstab`.
  - By using regular expression, we will gather tags of each question as list of strings.
  - Then, we will check whether any of these strings (tags) exist in the question title.
  - With the criteria above, we will create a new column in our dataframe and use this column along with Quality Rating information to create a crosstab.

```
import pandas as pd
from os.path import join

path_prefix = "./drive/My Drive"
```

```
fname = "stackoverflow.csv"
df = pd.read_csv(join(path_prefix, fname))
df.head()
```

```
# solution

import re

def is_in_title(question):
  """
  parses the tags and checks 
  if at least one of the tags is in the question title
  """
  title = question["Title"].lower()
  # find all the tags between arrow characters
  tags = re.findall(r"<([\w]+)>", question["Tags"])
  # check if at least one of the tags is in the question
  return any([tag.lower() in title for tag in tags])

# apply the function to all instances
# and create a new column named `tag_in_title`
# that stores boolean values
df["tag_in_title"] = df.apply(is_in_title, axis=1)
df.head()
```

```
# returns the crosstab between tag_in_title and Y columns
# displays the relationship between them
# here normalize parameter can be employed to work relative values
# normalize="index" -> normalization along rows
# normalize="columns" -> along columns
# normalize=True -> for all values
pd.crosstab(df["tag_in_title"], df["Y"], normalize="index")
```
