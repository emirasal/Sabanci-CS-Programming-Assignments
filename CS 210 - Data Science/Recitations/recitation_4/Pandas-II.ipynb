## Dealing with Missing Data

### Missing Value in Pandas: NaN

In computing, NaN, standing for Not a Number, is a member of a numeric data type that can be interpreted as a value that is undefined or unrepresentable. It is defined in the Numpy library as `np.nan` and Pandas primarily uses the value np.nan to represent missing data. It is by default not included in computations of Pandas functions.

Although `NaN` is similar to the native Python keyword `None`, there are some differences between the two. As you can observe below, while the data type of `np.nan` is float, an object with `None` keyword does has `NoneType`. If you are familiar with SQL *null* or JavaScript *undefined*, you can consider these as similar concepts.

Below, you can see the dataframe that we create to cover how to deal with missing data with Pandas. Note that, in addition to `np.nan`s we also have native Python keyword `None` to denote missing values. On top of these, some of the indices of the dataframe are filled with custom values like "*NA*" and "*Missing*" to also alert missing values. 

*We can consider the dataframe below as the dataset of our scenario where entries are the employees of a particular company. In our case, we need to submit this dataset to the registration system of the company and the super-boring-and-annoying system does not allow any missing data while submitting. We are in a rush to submit the information and we do not have the time to ask each individual to fill their information. Hence, we need to discover the ways of dealing with the missing data.*

Now, we'll see how to deal with them using Pandas.

````python
# "Missing", "NA" are custom missing values
# "None" is the native Python keyword for null
# "np.nan" denotes 'Not a Number' (aka. missing value) in Numpy, also mainly used in Pandas 

people = {
    'first': ['Adam', 'Jack', 'John', 'Chris', np.nan, None, 'NA'], 
    'last': ['Walker', 'Doe', 'Doe', 'Walker', np.nan, np.nan, 'Missing'], 
    'email': ['AdamWalker@gmail.com', 'JackDoe@email.com', 'JohnDoe@email.com', None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
````

````python
df = pd.DataFrame(people)
df
````

### Discarding Missing Values

*Imagine that we decided to remove the missing information from our dataset. We thought discarding rows with missing values could be straightforward, as we are in a rush to submit the information to the system.*

The main function to discard missing values in Pandas is the `dropna()` function. Here, we'll see various possible usages of this function.

Parameters that `dropna()` can take are also explained below.

    - how 
        ‘any’ : If any NA values are present, drop that row or column
        ‘all’ : If all values are NA, drop that row or column

    - axis
        0, or ‘index’ : Drop rows which contain missing values
        1, or ‘columns’ : Drop columns which contain missing value
   
    - subset
        Labels along other axis to consider, 
        e.g. if you are dropping rows these would be a list of columns to include.  

Information is taken from the corresponding [documentation page](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html).

### Groupby

An SQL-like feature is to apply `groupby` operation. It can be used to group a collection of data and compute operations on these groups.

````python
def assign_class(age):
    """
    age parameter will be coming from the values stored in "age" column
    """
    if age == 22:
        return "senior"
    elif age == 21:
        return "junior"
    elif age == 20:
        return "sophomore"
    else:
        return "freshman"
````

````python
data = {
    "age": np.random.choice(np.arange(18, 23), 10, p=[0.2, 0.1, 0.2, 0.1, 0.4]),
    "gpa": np.round(np.random.uniform(0, 4, (10, )), decimals=2),
    "name": np.random.choice(["luigi", "nikola", "diana", "cecilia"], 10)
}

df = pd.DataFrame(data)
df["class"] = df["age"].apply(assign_class)
df
````

Notice that using only `groupby` returns a *DataFrameGroupBy* (a groupby object) object.

````python
class_group = df.groupby(by="class")  # now, we only have a groupby object
class_group
````

````python
class_group.get_group("senior")
````

From this groupby object we can extract insights with a column selection first and then followed by applying a function.

````python
# In this case;
# class_group is a groupby object grouped already by the 'class' information
# then, 'age' column is selected from the groupby object -> this results rows with the same age values to be grouped
# with value_counts() -> we obtain the number of students with same 'class' and 'age' 

class_group["age"].value_counts()
````

````python
# we can also obtain aggregated statistics from the groupby object
# in this case -> we obtain mean ages for different classes in our dataset
class_group["age"].mean()
````

````python
df_new = pd.concat([df_a, df_b])  # merge first two dataframes
df_new
````

````python
pd.concat([df_a, df_b], ignore_index = True)
# Now, indices of the concatenated dataframe does not have repetitions. 
````

#### Merge

To achieve a database-style join, we need to supply the `merge` function with necessary parameters which specify the methods of merging operation between dataframes.

From these parameters;

  - **on:** defines the column name that dataframes will be joined on. The specified column should be present in both dataframes.
  - **how:** defines the type of the merge to be performed.


    Values that how parameter can take;
      left: use only keys from left frame, similar to a SQL left outer join.
      right: use only keys from right frame, similar to a SQL right outer join.
      outer: use union of keys from both frames, similar to a SQL full outer join.
      inner: use intersection of keys from both frames, similar to a SQL inner join.

````python
df_n
````

````python
df_new
````

````python
pd.merge(df_new, df_n, on='student_number')  # 'on' argument -> column or index level names to join on
````

## Working with String and Datetime Columns




### Datetime Objects

In Pandas, a dataframe can store values of different data types, such as float, strings and datetimes.

A `datetime` object helps you store and manipulate dates and times.

````python
from datetime import datetime
````

````python
# get current date
datetime.today()
````

````python
# datetime object with specified 
# year, month and day information
datetime(2017, 7, 25)
````

With the help of datetime objects, you can access different parts of a timestamp.

````python
d = datetime.today()
print("year:", d.year)
print("month:", d.month)
print("day:", d.day)
````

You can obtain more information for a specific datetime object. For instance, you can get the week day with the help of `weekday` attribute.

````python
# returns an integer between 0-6
# where 0 is Monday and 6 is Sunday
d.weekday()
````

In addition, you can perform datetime arithmetics as well.

````python
# returns a timedelta object
# and you can obtain the passed days
# with `days` attribute
days = datetime.today() - datetime(1900, 1, 1)
days.days
````

And lastly, you can convert strings of different formats into a datetime object. To this end, you can use a function named `strptime`, which takes the input string and the format as its parameters and returns a datetime object.

```py
>>> date = "01-01-2020"
>>> datetime.strptime(date, "%d-%m-%Y")
datetime.datetime(2020, 1, 1, 0, 0)
```

For the format, you need to specify the separators and locations of the meta-characters which stand for the datetime units, such as year, hour and seconds. In our example, we had `Day-Month-Year` as the format and to parse it we used `"%d-%m-%Y"`, which corresponds to the respective datetime units.

You can check the meta-character list from this [link](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior)

Likewise, you can also cast a datetime object into a string with a specified format with the help of `strftime` function.

```py
>>> d = datetime.today()
>>> d.strftime("%Y/%m")
'2020/11'
```

### Accessing String and Datetime Values from a DataFrame

Let's say we have the following toy dataset of students and their information. In the previous examples we use the same dataset without the `birth_date` column, which is just random date values.

When we check the dtypes of the columns, we can see that the `birth_date` column is of object data types; meaning that the values in there are nothing but strings.

````python
from datetime  import timedelta

data = {
    "gpa": np.round(np.random.uniform(0, 4, (10, )), decimals=2),
    "name": np.random.choice(["luigi", "nikola", "diana", "cecilia"], 10),
    "birth_date": [date.strftime("%d-%m-%Y") for date in pd.date_range(start="1/1/1995", end="31/12/2000", periods=10)]
}

df = pd.DataFrame(data)
df
````

````python
df.dtypes
````

First let's say that we want to capitalize the names stores in the `name` column. To do that we can create a function.

````python
def capitalize_names(name):
  return name.capitalize()

df["name"].apply(capitalize_names)
````

````python
df["name"][0]
````

````python
df["name"].capitalize()
````

It works; however, we wrote function for a simple string method. `str.capitalize()` is built-in string method. Things would be much easier if we could apply this method on the entire column. For instance

```py
>>> df["name"].capitalize()
'Series' object has no attribute 'capitalize'
```

However, this statement returns an error. Since a series object does not have a method named `capitalize`.

Fortunately, Pandas provides a useful accessor object for strings, which is named `str`. Whenever we append the accessor object to a series of objects, we obtain the string-like properties.

````python
df["name"].str
````

````python
# now we can apply capitalize the names
df["name"].str.capitalize()
````

Keep in mind that the returned value is again a series object.

Let's say you want to apply first `capitalize` and then remove `a`'s from the names. In that case, you need to call `str` accessor multiple times as needed.

````python
df["name"].str.capitalize().str.replace("a", "")
````

Datetime columns also have an accessor object named `dt`. However, in order to use `dt`, you need to make sure that the column you are working on is of datetime data type.

```py

```

So, we need to convert it to a datetime column. To do that, we can use the `to_datetime` function of Pandas.
